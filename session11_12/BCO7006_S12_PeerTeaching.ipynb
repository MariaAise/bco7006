## API 1 — REST Countries (no key)
`https://restcountries.com/v3.1/region/europe?fields=name,population,region`
Returns a **list** of country objects.

````python
try:
    import requests
    r = requests.get("https://restcountries.com/v3.1/region/europe",
                     params={"fields": "name,population,region"}, timeout=5)
    print("Status:", r.status_code)
    data = r.json()
    print("Top level is a", type(data).__name__, "of length", len(data))
    print("First entry keys:", list(data[0].keys()))
    print("Example:", data[0]["name"]["common"], "-", data[0]["population"])
except Exception as e:
    print("Run this in Colab with internet. (REST Countries offline: " + type(e).__name__ + ")")
````

## API 2 — Frankfurter exchange rates (no key)
`https://api.frankfurter.app/latest?from=USD`
Returns a **dict** with a nested `rates` dict.

````python
try:
    import requests
    r = requests.get("https://api.frankfurter.app/latest", params={"from": "USD"}, timeout=5)
    print("Status:", r.status_code)
    data = r.json()
    print("Top level is a", type(data).__name__, "with keys", list(data.keys()))
    print("USD -> EUR today:", data["rates"]["EUR"])
except Exception as e:
    print("Run this in Colab with internet. (Frankfurter offline: " + type(e).__name__ + ")")
````

## API 3 — ISS astronauts (no key)
`http://api.open-notify.org/astros.json`
Returns a **dict**; the people are in a nested **list** under `people`.

````python
try:
    import requests
    r = requests.get("http://api.open-notify.org/astros.json", timeout=5)
    print("Status:", r.status_code)
    data = r.json()
    print("People in space right now:", data["number"])
    print("First astronaut:", data["people"][0]["name"])
except Exception as e:
    print("Run this in Colab with internet. (Open Notify offline: " + type(e).__name__ + ")")
````

## API 4 — Store product categories (no key)
`https://fakestoreapi.com/products/categories`
Returns a **list of strings** — simpler than the others, a good first one.

````python
try:
    import requests
    r = requests.get("https://fakestoreapi.com/products/categories", timeout=5)
    print("Status:", r.status_code)
    data = r.json()
    print("Top level is a", type(data).__name__, ":", data)
except Exception as e:
    print("Run this in Colab with internet. (Fake Store categories offline: " + type(e).__name__ + ")")
````

# Session 8 — Pair Programming Activity
# Clean, Reshape, Combine
**Duration:** 40 min · **Pairs**

## Scenario
The marketing team wants a country-by-month revenue report. You have two raw files: `customers.csv` and `transactions.csv`. Build the analytical table they need.

## Working agreement
- Driver / Navigator. Swap each task.
- Run cells as you go — don't write 5 tasks of code before running any.

## Task 1 — Load and audit
Load both files and report:
- shape of each
- columns with missing values, and how many
- which columns are the wrong dtype (strings that should be numbers or dates)

## Task 2 — Fix missing customer ages
Decide a strategy (drop vs fill) and apply it. Document your decision in a comment.

**Discuss with your pair:** if 10% of customers had no age, would you still pick the same approach?

## Task 3 — Convert dates
- Convert `signup_date` (DD/MM/YYYY) and `txn_date` to real datetimes
- Add a `txn_month` column to transactions (use `.dt.to_period('M')`)

## Task 4 — Clean strings
- Strip whitespace from `email` and lowercase it
- Check `country` for any duplicates with different casing

## Task 5 — Join
Merge transactions with customers. Choose the join type carefully:
- We want EVERY transaction in the output
- Transactions without a matching customer should still appear (they're orphans worth investigating)

## Task 6 — Diagnose orphans
How many transactions have no matching customer? Which `customer_id` values?

## Task 7 — The analytical table
From your merged data (excluding orphans), build a pivot table with:
- **Rows:** country
- **Columns:** txn_month
- **Values:** sum of `amount`

## Task 8 — Top spenders per country
For each country, who are the top 2 customers by total spending?

**Hint:** `groupby` → `sum` → `sort_values` → `groupby('country').head(2)`

## Wrap up
- Save your final cleaned merged DataFrame to `clean_merged.csv`
- Be ready to show your pivot table

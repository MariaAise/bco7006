```python
import pandas as pd
import numpy as np
df = pd.read_csv('clean_merged.csv', parse_dates=['txn_date', 'signup_date'])
df['txn_month'] = df['txn_date'].dt.to_period('M')
df.head(2)
```

## Table 1 — Category performance

```python
t1 = (df.groupby('category')
        .agg(n_txns=('amount', 'count'),
             total_rev=('amount', 'sum'),
             avg_txn=('amount', 'mean'))
        .sort_values('total_rev', ascending=False)
        .round(2))
t1.to_csv('table1.csv')
t1
```

## Table 2 — Monthly trend

```python
t2 = df.pivot_table(index='txn_month', columns='category', values='amount',
                     aggfunc='sum', fill_value=0).round(2)
t2.to_csv('table2.csv')
t2
```

## Table 3 — Customer cohorts

```python
df['age_group'] = pd.cut(df['age'], bins=[0, 30, 45, 60, 200],
                          labels=['<30', '30-44', '45-59', '60+'])
t3 = (df.groupby('age_group', observed=True)
        .agg(n_customers=('customer_id', 'nunique'),
             total_spend=('amount', 'sum'),
             avg_txn=('amount', 'mean'))
        .round(2))
t3.to_csv('table3.csv')
t3
```

## Table 4 — Country × Category

```python
t4 = df.pivot_table(index='country', columns='category', values='amount',
                     aggfunc='sum', fill_value=0).round(2)
t4.to_csv('table4.csv')
t4
```

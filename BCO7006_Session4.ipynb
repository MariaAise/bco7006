# Part 1 — Conditional statements: `if` / `elif` / `else`

## Theory

Programs need to do different things depending on the data they're given. A bank app needs to react differently when your balance is positive vs. overdrawn. A pricing system applies different discounts to different customer tiers. The tool for this is the **conditional statement**.

The shape is:

```python
if some_condition:
    do_this
elif another_condition:
    do_that
else:
    do_this_otherwise
```

A few things to notice from the start:

- The **condition** is something that evaluates to `True` or `False` — exactly the comparison and logical operators you met in Session 3 (`>`, `==`, `and`, `or`, `not`).
- The line ends with a **colon** `:`. Forgetting it is the most common beginner error.
- The body is **indented** (4 spaces is conventional). Python uses indentation to know what belongs inside the `if`.
- `elif` and `else` are optional. You can have just `if`, or `if` + `else`, or any number of `elif` branches.

## 1.1 The simplest form: `if` alone

````python
# A single check — only print if the condition is True
order_total = 145

if order_total > 100:
    print("This is a premium order!")

print("Done.")
````

````python
# If the condition is False, the body is skipped entirely
order_total = 50

if order_total > 100:
    print("This is a premium order!")    # this line never runs

print("Done.")
````

## 1.2 `if` + `else`: handling both outcomes

If the answer to the question is "yes do A, otherwise do B", use `else`.

````python
order_total = 50

if order_total > 100:
    print("Premium order — free shipping!")
else:
    print("Standard order — $9.95 shipping.")
````

**Connecting back to Session 3:** the line `shipping = 0 if subtotal > 100 else 9.95` you used in last session's mini-project is just a compressed one-line version of an `if/else`. Now you can write it the long way:

```python
if subtotal > 100:
    shipping = 0
else:
    shipping = 9.95
```

Both forms produce the same result. The compressed version is fine for short cases; the long version is clearer for anything more complex.

## 1.3 `if` / `elif` / `else`: chained decisions

When you have **more than two** possible outcomes, chain `elif` (short for "else if") branches.

````python
# Customer tier classification — a classic BA pattern
lifetime_spend = 750

if lifetime_spend < 100:
    tier = "bronze"
elif lifetime_spend < 500:
    tier = "silver"
elif lifetime_spend < 2000:
    tier = "gold"
else:
    tier = "platinum"

print(f"Customer tier: {tier}")
````

Important rule: Python checks the conditions **in order, top to bottom**, and stops at the first one that's `True`. The order matters — if you wrote the bronze check (`< 100`) after silver (`< 500`), it would never run, because everything under 100 is also under 500.

## 1.4 Conditions can be anything that's True or False

Anything that produces a boolean works as a condition. That includes the logical combinations from Session 3:

````python
# Combining conditions with `and`, `or`, `not`
lifetime_spend = 750
is_premium_member = True
is_first_purchase = False

if lifetime_spend > 500 and is_premium_member and not is_first_purchase:
    print("Eligible for VIP discount")
else:
    print("Standard pricing")
````

### ⚠️ Watch out for the `=` vs `==` trap

The single most common beginner error in any programming language:

- `=` is **assignment** — `x = 5` means "store 5 in x"
- `==` is **comparison** — `x == 5` means "is x equal to 5?"

Mixing them up gives you either a `SyntaxError` or, worse, code that runs but does the wrong thing. If a condition isn't behaving as you expect, check this first.

# Part 2 — `for` loops

## Theory

Almost everything useful in business analytics involves doing the same thing to many pieces of data: total up a column of sales, filter customers by age, find the average price of products in a category. The tool for repeating work over a collection is the **`for` loop**.

The shape is:

```python
for variable_name in some_collection:
    do_something_with(variable_name)
```

In each pass through the loop, `variable_name` is set to the next item in the collection. The body runs, then it repeats with the next item, until there are no more items.

You already met this once. Last session, you copied this recipe:

```python
for k, v in some_dict.items():
    print(k, v)
```

Now you can read it: "for each key-value pair in the dictionary, set `k` to the key, set `v` to the value, and run the body."

## 2.1 The simplest case: looping over a list

````python
products = ["laptop", "mouse", "keyboard", "monitor"]

for product in products:
    print(f"Available: {product}")
````

The variable name (`product` here) is up to you — pick something that describes what each item *is*. Don't write `for x in products:` if `for product in products:` is clearer.

## 2.2 Looping a fixed number of times: `range()`

When you want the loop to run "5 times" or "from 1 to 10", use `range()`.

````python
# Loop 5 times — range(5) gives 0, 1, 2, 3, 4
for i in range(5):
    print(f"Iteration {i}")
````

````python
# range(start, stop) — stop is EXCLUSIVE
for i in range(1, 6):
    print(i)    # prints 1, 2, 3, 4, 5

print("---")

# range(start, stop, step) — count by 10
for i in range(0, 101, 10):
    print(i, end=" ")    # 0 10 20 30 ... 100
````

## 2.3 Pattern: the **accumulator**

A very common loop pattern: keep a running total (or count, or list) and update it each time through the loop.

````python
# Total up a list of prices
prices = [29.99, 15.50, 7.25, 49.00, 12.75]

total = 0                  # start the accumulator
for price in prices:
    total = total + price  # add each price to the running total

print(f"Total: ${total:.2f}")
````

````python
# Counting matches — count how many orders are above a threshold
order_amounts = [45, 120, 75, 300, 25, 180, 90]

count = 0
for amount in order_amounts:
    if amount > 100:
        count = count + 1   # also writeable as: count += 1

print(f"Number of large orders: {count}")
````

## 2.4 Pattern: the **filter**

Build a new list containing only the items that match a condition.

````python
# Find customers above a spending threshold
customers = [
    {"name": "Alice", "spend": 1200},
    {"name": "Bob",   "spend": 75},
    {"name": "Carol", "spend": 540},
    {"name": "Dave",  "spend": 30}
]

big_spenders = []          # start with an empty list
for customer in customers:
    if customer["spend"] > 500:
        big_spenders.append(customer["name"])

print("Big spenders:", big_spenders)
````

These two patterns — **accumulator** and **filter** — cover the vast majority of loops you'll write in business analytics. Get fluent with them and you're 80% there.

## 2.5 Looping over a dictionary, properly explained

Now that you know what a `for` loop is, here's the dict iteration recipe from Session 3 explained.

A dictionary has three views you can loop over:

- `dict.keys()` — just the keys
- `dict.values()` — just the values
- `dict.items()` — both, as a (key, value) pair

````python
catalogue = {"laptop": 999, "mouse": 25, "keyboard": 75, "monitor": 350}

print("--- Just keys ---")
for product_name in catalogue.keys():
    print(product_name)

print("--- Just values ---")
for price in catalogue.values():
    print(price)

print("--- Both ---")
for product_name, price in catalogue.items():
    print(f"{product_name}: ${price}")
````

The line `for product_name, price in catalogue.items():` is doing **two things at once**: getting the next pair from the dictionary, AND splitting that pair into two named variables. Python calls this **unpacking**, and it's a feature you'll see again.

### ⚠️ Watch out: don't change a list while you're looping over it

A surprisingly common bug. If you remove items from a list mid-loop, you'll skip items or hit weird errors. If you need to filter, build a *new* list (the filter pattern above) — don't modify the original.

# Part 3 — `while` loops (brief)

## Theory

A `for` loop is the right tool when you have a collection and want to do something to each item. But sometimes you don't have a collection — you just want to **keep going until some condition stops being true**. That's what `while` is for.

```python
while condition_is_true:
    do_something
```

The condition is checked **before each pass**. As soon as it becomes `False`, the loop ends and execution moves on.

The clearest rule of thumb: 
- *"I know exactly how many times / I have a collection"* → `for`
- *"I'll stop when something specific happens"* → `while`

## 3.1 A simple counting `while`

````python
# Equivalent of for i in range(5):
i = 0
while i < 5:
    print(f"i = {i}")
    i = i + 1   # IMPORTANT: must change `i` or the loop runs forever
````

## 3.2 Realistic use: input validation

Where `while` really earns its place: when you don't know how many tries it'll take.

````python
# Imagine prompting a user until they give a positive number
# (using a fake list of inputs so we don't actually prompt — but the shape is the same)

fake_inputs = [-5, 0, -2, 7]   # the user's attempts in order
attempt = 0

value = -1
while value <= 0:
    value = fake_inputs[attempt]
    attempt = attempt + 1
    print(f"User entered: {value}")

print(f"Accepted: {value}")
````

### ⚠️ Watch out: infinite loops

If the condition never becomes `False`, the loop runs forever. In Colab this means a hung cell. The two ways this happens most often:

1. You forgot to update the variable inside the loop (e.g. forgetting `i = i + 1`).
2. The update can never make the condition false (e.g. `while x > 0: x = x + 1`).

If you ever hit this in Colab, click the stop button at the top of the cell.

# Part 4 — Functions

## Theory

You've been writing code that does things. Functions are how you **package up** that code, give it a name, and reuse it.

Why does this matter? Imagine your Session 3 mini-project asked for a discount calculation in three different places. You'd copy and paste it three times. Now imagine the discount rule changes — you have to find and fix all three copies, and pray you don't miss one. Functions solve this: write the rule **once**, give it a name, call it three times.

The shape:

```python
def function_name(parameter1, parameter2):
    # body — does the work
    return some_result
```

Four key terms to learn now (you'll use them constantly):

- **`def`** — the keyword that starts a function definition
- **Parameter** — a name in the function definition that stands for an input value (e.g. `parameter1`)
- **Argument** — the actual value you pass when you call the function
- **`return`** — sends a value back to whoever called the function

## 4.1 The simplest function — no inputs, no return

````python
# Define
def say_hello():
    print("Hello!")

# Call (use)
say_hello()
say_hello()
say_hello()
````

**Anatomy:**
- `def say_hello():` — defines a function called `say_hello` that takes no parameters
- The indented body runs each time the function is called
- `say_hello()` — the parentheses *call* (run) the function

A definition without a call does nothing — Python just remembers the function for later.

## 4.2 Functions with parameters — taking inputs

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
greet("Carol")
````

When you call `greet("Alice")`, Python:
1. Sets the parameter `name` to `"Alice"`
2. Runs the body
3. When the body ends, control returns to where you called the function

You can have as many parameters as you need:

````python
def greet_with_role(name, role):
    print(f"Hello, {name}! Welcome to your role as {role}.")

greet_with_role("Alice", "data analyst")
greet_with_role("Bob", "BA student")
````

## 4.3 `return` — sending a value back

Most useful functions don't just print things; they **compute** something and return it. You then capture the result in a variable.

````python
def double(x):
    return x * 2

result = double(7)
print(result)        # 14

# You can use the returned value directly in expressions
print(double(5) + double(10))   # 30
````

````python
# Realistic example — compute price after tax
def add_tax(price, tax_rate):
    return price * (1 + tax_rate)

print(add_tax(100, 0.10))      # 110.0
print(add_tax(49.99, 0.10))    # 54.989...
````

**Important:** the moment `return` runs, the function exits. Anything after `return` in the function body never executes.

A function without an explicit `return` automatically returns a special value called `None` (Python's "nothing" value). That's fine for functions that just print; functions that compute should always return.

## 4.4 Default arguments

Sometimes a parameter usually has the same value, but you want to allow overriding. Provide a **default** in the function definition.

````python
def add_tax(price, tax_rate=0.10):    # tax_rate defaults to 0.10
    return price * (1 + tax_rate)

# Use the default
print(add_tax(100))         # 110.0

# Override the default
print(add_tax(100, 0.20))   # 120.0
print(add_tax(100, tax_rate=0.05))   # 105.0 — naming makes it clearer
````

Rule: parameters with defaults must come **after** parameters without defaults in the function definition.

## 4.5 Functions that work on dictionaries

This pattern matters a lot — pay attention. A function can take a dictionary as a parameter and use its fields:

````python
def describe_product(product):
    name = product["name"]
    price = product["price"]
    stock = product["stock"]
    return f"{name} costs ${price} ({stock} in stock)"

laptop = {"name": "laptop", "price": 999, "stock": 15}
mouse  = {"name": "mouse",  "price": 25,  "stock": 200}

print(describe_product(laptop))
print(describe_product(mouse))
````

Notice what just happened: **one function** can describe **any** product, as long as the dictionary has the right shape. This is the seed of a powerful idea — and you'll see it bloom into "objects" in Session 5.

## 4.6 Functions that build dictionaries

Equally common: take some inputs, return a structured dictionary.

````python
def build_customer(name, email, city):
    return {
        "name": name,
        "email": email,
        "city": city,
        "lifetime_spend": 0    # new customers start at zero
    }

alice = build_customer("Alice Wong", "alice@example.com", "Melbourne")
bob   = build_customer("Bob Singh",   "bob@example.com",   "Sydney")

print(alice)
print(bob)
````

## 4.7 A note on variable scope (briefly)

A variable created **inside** a function only exists inside that function. Once the function returns, those variables are gone. This is called **local scope** — and it's a feature, not a bug. It means you can use the same name `x` in two different functions without them interfering.

````python
def calculate_thing():
    x = 10              # local to this function
    return x * 2

result = calculate_thing()
print(result)
# print(x)            # This would crash — x doesn't exist out here
````

# Part 5 — Reading text files

## Theory

Until now your data has lived inside the notebook — typed straight into the code. Real data doesn't work that way: it lives in **files** (`.csv`, `.txt`, `.json`, etc.), often produced by other programs or downloaded from systems.

Today you'll learn to read a simple text file. Specifically, the most common shape: each line is one record, with values separated by commas (a "CSV-like" file).

## 5.1 The `with open(...)` pattern

This is the recommended way to read a file in Python:

```python
with open("filename.txt") as f:
    contents = f.read()
```

Three things to notice:
- `open(...)` opens the file
- `as f` gives you a name (`f`) to refer to the open file
- `with` makes sure the file is **closed properly** when the block ends (even if something goes wrong)

The variable name `f` is conventional — short for "file".

## 5.2 First, let's create a sample file to read

In Colab you can create a file from inside the notebook:

````python
# Create products.csv with some sample data
csv_content = '''name,price,stock
laptop,999,15
mouse,25,200
keyboard,75,80
monitor,350,30
webcam,80,50
cable,12,400'''

with open("products.csv", "w") as f:
    f.write(csv_content)

print("File written.")
````

(The `"w"` in `open(..., "w")` means **write mode**. Without it, `open()` defaults to **read mode** — which is what you want most of the time.)

## 5.3 Reading the contents

````python
# Read the whole file as one big string
with open("products.csv") as f:
    contents = f.read()

print(contents)
print("---")
print(f"Type: {type(contents).__name__}")
print(f"Length: {len(contents)} characters")
````

`.read()` gives you everything as a single string. That's not directly useful — we want to work with one **line** at a time.

## 5.4 Reading line by line

````python
with open("products.csv") as f:
    lines = f.read().splitlines()

print(lines)
print(f"Type: {type(lines).__name__}, with {len(lines)} items")
````

`.splitlines()` is a string method that breaks a string into a list of lines (and strips the newline characters for you). Now we have a list of strings — and we know how to loop over those.

## 5.5 Parsing each line into a dictionary

The first line is usually a **header** with the column names. The remaining lines are data.

````python
with open("products.csv") as f:
    lines = f.read().splitlines()

header = lines[0].split(",")     # ["name", "price", "stock"]
print("Header:", header)

# Build a list of product dictionaries
products = []
for line in lines[1:]:           # skip the header (start at index 1)
    values = line.split(",")
    product = {
        "name":  values[0],
        "price": int(values[1]),    # convert text "999" to integer 999
        "stock": int(values[2])
    }
    products.append(product)

# Show what we built
for p in products:
    print(p)
````

Two important details from that code:

1. **`lines[1:]`** — slicing from index 1 to the end skips the header row.
2. **`int(values[1])`** — values from a file always come in as **strings**. If you want to do arithmetic, you must convert. Use `int()` for whole numbers, `float()` for decimals.

You now have a list of product dictionaries — exactly the shape we used in Session 3, but built from a file rather than typed by hand. That's a big step.

### What's next — Session 5

Look back at your mini-project. You wrote functions that kept passing around the same `customer`, `cart`, and `catalogue` dictionaries:

```python
compute_subtotal(cart, catalogue)
get_discount_rate(tier)
process_order(customer, cart, catalogue)
```

That's a sign that the data and the operations on that data are conceptually one thing. **Object-oriented programming** lets you bundle them together. Instead of:

```python
process_order(customer, cart, catalogue)
```

you'll write:

```python
customer.process_order(cart)
```

Same logic, but the customer and the operations that work on a customer live in the same place. You'll learn to define your own types of things (classes), create instances of them, and give them methods.

This is also how Python itself is built — every list, string, and dictionary you've used is an object with methods. `"hello".upper()` is a method call. You'll see it everywhere once you know to look for it.

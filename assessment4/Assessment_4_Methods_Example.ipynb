## Step 1 — The code cell

Below is a typical data-loading-and-cleaning cell. Read it, then read the explanation underneath. This is the level of explanation we expect for each data-handling cell in your appendix.

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("train.csv")

# Quick look at size and missing values
print("Rows, columns:", df.shape)
missing = df.isnull().sum()
print(missing[missing > 0])

# Fill missing numeric values with the column average
for col in df.columns:
    if df[col].dtype in ["float64", "int64"]:
        df[col] = df[col].fillna(df[col].mean())
    else:
        df[col] = df[col].fillna("Unknown")

# Flag rows where the price is unusually high (above the 95th percentile)
threshold = np.percentile(df["SalePrice"], 95)
df["high_value"] = df["SalePrice"] > threshold

df.head()
```

## Step 2 — The explanation (this is the gold standard)

**What it does.** Loads the dataset, reports its size and which columns have missing values, fills those gaps, and adds a column flagging unusually high-priced rows.

**Libraries used.**
- `pandas` (imported as `pd`) — used to load the data and work with it as a table (a *DataFrame*). Every `df.` call comes from pandas.
- `numpy` (imported as `np`) — used here for the numerical calculation of a percentile.

**pandas / numpy methods, line by line.**
- `pd.read_csv("train.csv")` — reads the CSV file into a DataFrame called `df`.
- `df.shape` — returns the number of rows and columns, so we can confirm the data loaded correctly.
- `df.isnull().sum()` — `isnull()` marks every empty cell as True/False; `.sum()` then counts those per column, giving the number of missing values in each column.
- `df[col].dtype` — returns the data type of a column, so we can tell numbers apart from text.
- `df[col].fillna(...)` — replaces missing values; with `df[col].mean()` for numeric columns (the column average) and with `"Unknown"` for text columns.
- `np.percentile(df["SalePrice"], 95)` — returns the value below which 95% of prices fall (the 95th percentile).

**The loop.** The `for col in df.columns:` loop repeats once for every column in the table. For each column it decides how to fill missing values.

**The condition.** The `if df[col].dtype in [...]` line decides *how* to fill: if the column holds numbers, fill with the average; `else` (text columns), fill with the word "Unknown". The second condition, `df["SalePrice"] > threshold`, produces True for each row priced above the threshold and False otherwise — that True/False result becomes the new `high_value` column.

**Why it matters for the business.** This cell makes the data trustworthy before any analysis: nothing is missing, and high-value records are easy to isolate.

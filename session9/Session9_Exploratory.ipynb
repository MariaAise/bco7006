## 1. Array creation patterns

```python
import numpy as np

# eye — identity matrix
print(np.eye(3))
```

```python
# logspace — like linspace but logarithmic
print(np.logspace(0, 3, 4))  # 10^0, 10^1, 10^2, 10^3
```

```python
# meshgrid — generate coordinate grids (useful for evaluating functions over a 2D space)
x = np.linspace(-2, 2, 5)
y = np.linspace(-2, 2, 5)
X, Y = np.meshgrid(x, y)
print("X:")
print(X)
print("\nY:")
print(Y)
print("\nZ = X² + Y²:")
print(X**2 + Y**2)
```

## 2. Advanced indexing

```python
arr = np.array([10, 20, 30, 40, 50])

# Fancy indexing — pass an array of indices
idx = [0, 2, 4]
print(arr[idx])  # [10, 30, 50]

# Even works with repeats
print(arr[[0, 0, 1, 1, 4]])  # [10, 10, 20, 20, 50]
```

```python
# 2D fancy indexing
m = np.arange(20).reshape(4, 5)
print(m)
# Get specific rows
print("\nRows 0 and 3:")
print(m[[0, 3]])
```

## 3. Reshape, stack, split

```python
a = np.arange(6)
print("Original:", a)
print("\nAs column vector (shape 6,1):")
print(a.reshape(-1, 1))
print("\nAs row vector (shape 1,6):")
print(a.reshape(1, -1))
```

```python
# Stacking
top = np.array([1, 2, 3])
bot = np.array([4, 5, 6])
print("vstack (vertical):")
print(np.vstack([top, bot]))
print("\nhstack (horizontal):")
print(np.hstack([top, bot]))
print("\ncolumn_stack:")
print(np.column_stack([top, bot]))
```

## 4. Broadcasting deep dive

```python
# Normalising columns of a matrix — subtract each column's mean
m = np.array([[10, 100, 1000],
              [20, 200, 2000],
              [30, 300, 3000]], dtype=float)

col_means = m.mean(axis=0)
print("Column means:", col_means)

normalised = m - col_means  # broadcasts row-wise
print("\nNormalised:")
print(normalised)
```

```python
# Now normalising ROWS — subtle: need to reshape the means to a column
row_means = m.mean(axis=1)
print("Row means:", row_means)
print("Shape:", row_means.shape)

# Won't broadcast as-is. Need a column shape.
row_means_col = row_means.reshape(-1, 1)
print("\nReshaped to column:")
print(row_means_col)

print("\nRow-normalised:")
print(m - row_means_col)
```

## 5. Random sampling for simulation

```python
# Coin flip simulation — what's the chance of getting >55 heads in 100 flips?
np.random.seed(0)
trials = 100_000
flips = np.random.randint(0, 2, size=(trials, 100))  # 0 or 1
heads_per_trial = flips.sum(axis=1)
prob = (heads_per_trial > 55).mean()
print(f"P(>55 heads in 100 flips) ≈ {prob:.4f}")
```
